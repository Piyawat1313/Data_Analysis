## ทำความรู้จักกับชุดข้อมูล load_wine และเติมคำตอบลงในใบงานส่วนที่ 1 

```python
from sklearn.datasets import load_wine
import pandas as pd
data = load_wine()

X = data.data
Y = data.target
print(X)
print(Y)
print(data.target_names)
```

## Target ของชุดข้อมูลนี้มี 

```python
print(data.target_names)
```

## Target แต่ละประเภท มีข้อมูลกี่ Samples

```python
import numpy as np

classes, count = np.unique(data.target, return_counts=True)
for c, cu in zip(classes, count):
    print(c, cu);
```

## แยกข้อมูล Train 80% และ Test 20% (ใช้คำสั่ง train_test_split 

```python
from sklearn.model_selection import train_test_split
from sklearn import tree


X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, train_size=0.80, random_state=0)

model = tree.DecisionTreeClassifier(max_depth=1)

model = model.fit(X_train, Y_train)

X_predic = model.predict(X_test)
```

## Predict Model 

```python
model = tree.DecisionTreeClassifier(max_depth=1)

model.fit(X_train, Y_train)

X_predic = model.predict(X_test)
```

## หาค่า Accuracy

```python
from sklearn.metrics import precision_score

scores = model.score(X_test, Y_test)
print(scores)
```

## หาค่า Precision

```python
from sklearn.metrics import precision_score

precission = precision_score(Y_test, X_predic, average='macro')
print(precission)
```

## หาค่า Recall

```python
from sklearn.metrics import recall_score

recall = recall_score(Y_test, X_predic, average='macro')
print(recall)
```

## หาค่า F1 score

```python
from sklearn.metrics import f1_score

f1 = f1_score(Y_test, X_predic, average='macro')
print(f1)
```

## จงเขียน Confusion Matrix ของผลการทดลองที่ดีที่สุด โดยระบุด้วยว่าเลือก max_depth เท่ากับเท่าไหร่ ระบุ Accuracy, Precision, Recall และ F1-score ลงในพื้นที่ว่างนี้ (เนื่องจากโจทย์ข้อนี้มีจำนวนคลาสมากกว่า 2 คลาส (คลาส = ประเภท) ดังนั้นการหาค่า TP, TN, FP และ FN จึงมีความซับซ้อนกว่าแบบ 2 คลาส โดยจำเป็นต้องเปลี่ยนวิธีการคำนวณ Average ใหม่ ปัญหาต่อมาคือ ควรใช้วิธีการ Average แบบไหนจึงจะเหมาะสม) 

```python
from sklearn.metrics import confusion_matrix

metrix = confusion_matrix(Y_test, X_predic)
print(metrix)
```

```python
max_d = [1,2,3,4,5,10,100,1000]
for i in max_d:
    X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, train_size=0.80, random_state=0)

    model = tree.DecisionTreeClassifier(max_depth=i)

    model = model.fit(X_train, Y_train)

    X_predic = model.predict(X_test)
    
    scores = model.score(X_test, Y_test)
    print(f"i = {i} Acc = {scores}")

    precission = precision_score(Y_test, X_predic, average='macro')
    print(f"i = {i}  precission = {precission}")

    recall = recall_score(Y_test, X_predic, average='macro')
    print(f"i = {i}  recall = {recall}")

    f1 = f1_score(Y_test, X_predic, average='macro')
    print(f"i = {i}  f1 = {f1}")
```
