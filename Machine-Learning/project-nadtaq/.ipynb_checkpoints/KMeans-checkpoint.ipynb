```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

# Cleaning Data
#### วิธีทำ
1. Load data
2. Data Reshaping (Pivot Table)
3.  Return Transformation (Daily Log Returns)
4.  Handling Missing Values (ตัดหุ้นที่มีค่าว่างออก)
5.  Standardization & Transpose (ปรับสเกลและสลับมิติให้หุ้นเป็น Samples พร้อมเข้าโมเดล)

```python
# step 1 
df = pd.read_csv('NASDAQ-100/NASDAQ_100_Data_From_2010.csv', sep='\t')
df['Date'] = pd.to_datetime(df['Date'])

# step 2
price_matrix = df.pivot(index='Date', columns='Name', values='Adj Close')

#step 3 
returns_matrix = np.log(price_matrix / price_matrix.shift(1)).iloc[1:]

# step 4 
clean_returns = returns_matrix.dropna(axis=1)

# step 5 
scaler = StandardScaler()
# X = scaler.fit_transform(clean_returns)
X = scaler.fit_transform(clean_returns).T


stock_names = clean_returns.columns.tolist()
print(stock_names)
print("จำนวนหุ้น: ", len(stock_names))
print(f"ขนาดของข้อมูล: {X.shape}")
```

# KMeans 
#### ขั้นตอนการทำ
1. หาค่า K ก่อน
   - ใช้วิธี Elbow Method เพื่อหาจุดหักศอก
3. กำหนดโมเดลและจำนวน K
4. ให้โมเดลเรียนรู้และจัดกลุ่มข้อมูล
5. นำผลลัพธ์ไปใช้งาน

```python
# step 1.1
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia = []
K_range = range(1, 10)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)


# กราฟ Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o', color='blue', linewidth=2)
plt.title('Elbow Method (inertia)')
plt.xlabel('Number of cluster (K)')
plt.ylabel('inertia')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(K_range)
plt.tight_layout()
plt.show()
```

## KMean เมื่อได้ค่า K เรียบร้อยแล้ว
1. train model KMean
2. ทำนายข้อมูลและดึงจุด Centroids
3. ดึงค่า Centroids ซึ่งมีความยาวเท่ากับจำนวนวัน
4. plot Graph


## องค์ประกอบของแกน
- แกน X (Time (Days)): ระยะเวลาเป็นจำนวนวัน ตั้งแต่วันที่ 0 ถึงเกือบ 3,000 วัน
- แกน Y (Normalized Return): ผลตอบแทนเฉลี่ยรายวันที่ถูก Standardize/Normalize แล้ว
  - ค่าใกล้ 0: ผลตอบแทนเปลี่ยนแปลงปกติ/นิ่งๆ
  - ค่าเป็น + สูงๆ: วันที่หุ้นกลุ่มนั้นปรับตัวขึ้นแรงผิดปกติ
  - ค่าเป็น - สูงๆ: วันที่หุ้นกลุ่มนั้นปรับตัวลดลงแรงผิดปกติ (Crash)
- เส้นแต่ละสี (Cluster 1-4): แทนค่าเฉลี่ยพฤติกรรมราคาหุ้นของแต่ละกลุ่ม


```python
# step 1
kmean = KMeans(n_clusters=4, random_state=42, n_init=10)
kmean.fit(X)

# step 2
y_kmean = kmean.predict(X)
centroids = kmean.cluster_centers_

# step 3
plt.figure(figsize=(7, 5))
for i in range(kmean.n_clusters):
    plt.plot(centroids[i], label=f'Cluster {i + 1}')

# step 4
plt.title('Average Daily Log Returns Pattern per Cluster')
plt.xlabel('Time (Days)')
plt.ylabel('Normalized Return')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## กราฟแสดงแถบสีของแต่ละ Cluster


### การดูกราฟ
1. สังเกตความผันผวนเฉพาะช่วงเวลา (Volatility Spikes)
  - ช่วงวันประมาณ 2,500 - 2,600:
    - เกิด Spike ขนาดใหญ่มาก ทั้งฝั่งบวกและลบ โดยเฉพาะ Cluster 2 (สีส้ม) และ Cluster 4 (สีแดง) ดิ่งลงไปถึงเกือบ-8
    - การตีความ: ช่วงเวลานี้เกิดเหตุการณ์วิกฤต หรือ Market Shock บางอย่างที่ส่งผลกระทบต่อหุ้นใน Cluster 2 และ 4 อย่างรุนแรงเป็นพิเศษ

2. สังเกตความต่างของพฤติกรรมระหว่าง Cluster
   - Cluster 1 (สีฟ้า) & Cluster 3 (สีเขียว):
     - มีการแกว่งตัวผันผวนประปรายตลอดช่วงเวลา (เช่น ช่วงวัน 800 หรือ 1,700) แต่ไม่ดิ่งลบแรงเท่า Cluster 2 และ 4 ในช่วงวัน 2,500
    - Cluster 2 (สีส้ม) & Cluster 4 (สีแดง):
       - มักจะเคลื่อนไหวไปในทิศทางเดียวกันในหลายช่วงเวลา แต่ Cluster 2 จะมีความผันผวนสุดขั้ว (Extreme Outliers) มากกว่า


### สรุปสิ่งที่กราฟนี้บอกเรา
1. หุ้นถูกจัดกลุ่มตาม "ระดับความเสี่ยง/ความผันผวนในวันเกิดวิกฤต": อัลกอริทึมมองเห็นว่ามีหุ้นบางกลุ่มที่ทนทานต่อแรงกระแทกได้ดีกว่า และหุ้นบางกลุ่มที่ผันผวนรุนแรงมากเมื่อเกิดเหตุการณ์ไม่ปกติ

```python
fig, axes = plt.subplots(4, 1, figsize=(14, 10), sharex=True, sharey=True)

colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

for i in range(4):
    axes[i].plot(centroids[i], color=colors[i], label=f'Cluster {i+1}')
    axes[i].set_title(f'Cluster {i+1} Return Pattern')
    axes[i].grid(True, linestyle='--', alpha=0.5)
    axes[i].set_ylabel('Normalized Return')

plt.xlabel('Time (Days)')
plt.tight_layout()
plt.show()
```

## KMean ร่วมกับ Gaussian Distribution
### วิธีทำ
1. กำหนดจำนวน K = 4 ตามที่คุณเลือกจาก Elbow Method
2. สร้าง GMM โดยตั้งค่า init_params='kmeans' เพื่อใช้ K-Means หาจุดเริ่มต้น
3. เรียนรู้และทำนายกลุ่ม
4. ดึงค่า Probability (ความน่าจะเป็นที่หุ้นแต่ละตัวจะอยู่ในแต่ละ Cluster)
5. สร้าง DataFrame สรุปว่าหุ้นไหนอยู่ Cluster ไหน

```python
from sklearn.mixture import GaussianMixture
# step 1 
n_clusters = 4

# step 2 
gmm = GaussianMixture(
    n_components=n_clusters,
    init_params='kmeans', # ใช้ K-Means กำหนดจุดเริ่มต้น
    covariance_type='full', # อนุญาตให้ Covariance กระจายตัวได้ทุกทิศทาง (รูปวงรี)
    random_state=42
)

# step 3
cluster_labels = gmm.fit_predict(X)

# step 4
cluster_probs = gmm.predict_proba(X)

# step 5 
df_clusters = pd.DataFrame({
    'Stock': stock_names,
    'Cluster': cluster_labels
})

df_clusters.head(10)
```

## คัดเลือกหุ้นตัวแทนกลุ่ม พิจารณาจาก
1. หุ้นที่มีความน่าจะเป็นสูงที่สุดในการเป็นสมาชิกของ Cluster นั้นๆ
2. หุ้นที่มี Sharpe Ratio สูงสุดใน Cluster นั้นๆ

### ขั้นตอนการทำ
1. คำนวณ Sharpe Ratio รายวันของหุ้นแต่ละตัว (สมมติ Risk-Free Rate = 0)
2. เลือกหุ้นที่มี Sharpe Ratio สูงที่สุดในแต่ละ Cluster เพื่อมาเป็นตัวแทนทีม

```python
# step 1 
daily_mean = clean_returns.mean()
daily_std = clean_returns.std()
sharpe_ratio = (daily_mean / daily_std) * np.sqrt(252)

df_clusters['Sharpe_Ratio'] = df_clusters['Stock'].map(sharpe_ratio)

# step 2 
selected_stocks = []
for cluster in range(n_clusters):
    cluster_stocks = df_clusters[df_clusters['Cluster'] == cluster]
    best_stock = cluster_stocks.sort_values(by='Sharpe_Ratio', ascending=False).iloc[0]['Stock']
    selected_stocks.append(best_stock)

print("หุ้นตัวแทนที่ได้รับคัดเลือกเข้าพอร์ต (Selected Team):", selected_stocks)
```

## จัดสรรน้ำหนักพอร์ตโฟลิโอ
- นำหุ้นตัวแทนที่คัดเลือกมาคำนวณหา Optimal Weights เพื่อสร้างพอร์ตที่มีความเสี่ยงต่ำที่สุดหรือให้ Sharpe Ratio สูงที่สุด

### ขั้นตอนการทำ
1. ดึงข้อมูล Return เฉพาะหุ้นที่ได้รับคัดเลือก
2. คำนวณ Mean Return และ Covariance Matrix (อิงหลักการ Gaussian Distribution)
3. ฟังก์ชันคำนวณ ค่าความผันผวนรวม และ Sharpe Ratio ของพอร์ต
4. ฟังก์ชันเป้าหมาย (Objective Function) เพื่อหา Max Sharpe Ratio (Minimize -Sharpe)
5. เงื่อนไข: ผลรวมน้ำหนักต้องเท่ากับ 1 (100%) และน้ำหนักแต่ละตัวอยู่ระหว่าง 0 ถึง 1
6. Optimize
7. ผลลัพธ์น้ำหนักการลงทุนที่เหมาะสมที่สุด

```python
from scipy.optimize import minimize

# step 1 
portfolio_returns = clean_returns[selected_stocks]

# step 2 
mean_returns = portfolio_returns.mean() * 252
cov_matrix = portfolio_returns.cov() * 252 

# step 3 
def portfolio_performance(weights, mean_returns, cov_matrix):
    returns = np.dot(weights, mean_returns)
    std_dev = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))
    sharpe = returns / std_dev
    return std_dev, sharpe

# step 4 
def negative_sharpe(weights, mean_returns, cov_matrix):
    return -portfolio_performance(weights, mean_returns, cov_matrix)[1]

# step 5 
num_assets = len(selected_stocks)
constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
bounds = tuple((0, 1) for _ in range(num_assets))
initial_weights = num_assets * [1. / num_assets]

# step 6 
opt_results = minimize(
    negative_sharpe,
    initial_weights,
    args=(mean_returns, cov_matrix),
    method='SLSQP',
    bounds=bounds,
    constraints=constraints,
)

# step 7 
optimal_weights = opt_results.x

print("\n--- น้ำหนักการลงทุนที่เหมาะสมที่สุด (Optimal Weights) ---")
for stock, weight in zip(selected_stocks, optimal_weights):
    print(f"หุ้น {stock}: {weight*100:.2f}%")
```

# DBSCAN
### ขั้นตอนการทำ
1. คำนวณ Correlation Matrix -> Distance Matrix
2. แปลง Correlation เป็น Distance (ระยะห่างระหว่างหุ้น)
   - ค่า Distance จะอยู่ระหว่าง 0 (เหมือนกันเป๊ะ) ถึง 2 (ตรงข้ามกันเป๊ะ)
3. หาค่า eps ด้วย k-distance graph
   - กำหนดจำนวนสมาชิกขั้นต่ำในกลุ่ม
4. คำนวณระยะห่างไปยังเพื่อนที่ใกล้ที่สุดลำดับที่ k (k = min_samples)
5. เรียงลำดับระยะห่างเพื่อวาด Elbow Plot

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

# step 1 
corr_matrix = clean_returns.corr()

# step 2 
dist_matrix = np.sqrt(2 * ( 1 - corr_matrix))

# step 3 
min_samples = 3 

# step 4 
neighbors = NearestNeighbors(n_neighbors=min_samples, metric='precomputed')
neighbors_fit = neighbors.fit(dist_matrix)
distances, indices = neighbors_fit.kneighbors(dist_matrix)

# step 5 
sort_distances = np.sort(distances[:, min_samples - 1])

plt.figure(figsize=(8, 4))
plt.plot(sort_distances)
plt.title('k-distance Graph for Finding Optimal eps')
plt.xlabel('Points sorted by distance')
plt.ylabel(f'{min_samples}-NN Distance')
plt.grid(True)
plt.show()
```

```python
optimal_eps = 0.80

dbscan = DBSCAN(eps=optimal_eps, min_samples=min_samples, metric='precomputed')
db_labels = dbscan.fit_predict(dist_matrix)

# สร้าง DataFrame ผลลัพธ์
df_dbscan = pd.DataFrame({
    'Stock': clean_returns.columns,
    'Cluster': db_labels
})

# สรุปผลการจัดกลุ่ม
print("--- จำนวนหุ้นในแต่ละ Cluster ---")
print(df_dbscan['Cluster'].value_counts().sort_index())
```
