```python
from sklearn.linear_model import Ridge
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

# สร้างข้อมูล และ ดึงชื่อของ Target เอามาแสดงผล

```python
data = load_diabetes()
x = data.data
y = data.target
print(x.shape)
print(data.feature_names)
print(f"x = {x}")
print(f"y = {y}") 
```

# ใช้ Model Ridge ทำนายและหาค่า mean_squared_error

```python
test001 = [0.01, 0.05, 0.20, 0.80, 0.95, 0.99]
for i in test001:
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=i, random_state=0)
    model = Ridge(alpha=0.1) 
    model.fit(x_train, y_train)
    pred = model.predict(x_test)
    mse = mean_squared_error(y_test, pred)
    print(mse)
```

# ใช้ Model LinearRegreession ในการทำนายข้อมูลและหาค่า mean_squared_error

```python
from sklearn.linear_model import LinearRegression
```

```python
for i in test001:
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=i, random_state=0)
    model = LinearRegression()
    model.fit(x_train, y_train)
    predic = model.predict(x_test)
    mse = mean_squared_error(y_test, predic)
    print(f"mean_squared_error = {mse}")
```

# ใช้ Model Lasso ทำนายข้อมูลและหาค่า mean_squared_error

```python
from sklearn.linear_model import Lasso
```

```python
for i in test001:
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = i, random_state = 0)
    model = Lasso(alpha = 0.1)
    model.fit(x_train, y_train)
    pred = model.predict(x_test)
    mse = mean_squared_error(y_test, pred)
    print(f"test size = {i} mean_squared_error = {mse}")
```

```python
from sklearn.linear_model import ElasticNet
```

```python
for i in test001:
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = i, random_state = 0)
    model = ElasticNet(alpha = 0.01)
    model.fit(x_train, y_train)
    pred = model.predict(x_test)
    mse = mean_squared_error(y_test, pred)
    print(f"test size = {i} mean_squared_error = {mse}")
```
