```python
from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import mean_absolute_error
import numpy as np
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import r2_score
```

```python
data = load_linnerud()
x = data.data
y = data.target

x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42, test_size=0.3)
```

```python
list_lr = [0.001, 0.005, 0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
mae_s = []
r2 = []

for i in list_lr:
    model_gra = MultiOutputRegressor(GradientBoostingClassifier(learning_rate=i, random_state=42)) 
    model_gra.fit(x_train, y_train)
    y_pred = model_gra.predict(x_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    mae_s.append(mae)

    r2_point = r2_score(y_test, y_pred)
    r2.append(r2_point)

for i in mae_s:
    print(i)

print("\n===== R2_scores ===========")
for i in r2:
    print(i)
```

```python
n_estimator = [1,5,10,20,50,100,200,500]
mae_s2 = []
r22 = []
lear_ning = 0.001

for i in n_estimator:
    model_gra = MultiOutputRegressor(GradientBoostingClassifier(learning_rate=lear_ning, random_state=42, n_estimators=i)) 
    model_gra.fit(x_train, y_train)
    y_pred = model_gra.predict(x_test)
    
    mae2 = mean_absolute_error(y_test, y_pred)
    mae_s2.append(mae2)

    r2_point1 = r2_score(y_test, y_pred)
    r22.append(r2_point1)

print("==== MAE ====") 
for i in mae_s2:
    print(i)

print("==== R2 ====")
for i in r22:
    print(i)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_wine
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
```

```python
lw = load_wine()
x1 = lw.data
y1 = lw.target

x_train1, x_test1, y_train1, y_test1 = train_test_split(x1, y1, test_size=0.2, random_state=42)
```

```python
max2 = [1,2,3,4,5,10,100,1000]
acc_list = []
f1_list = []
for i in max2:
    rfc = RandomForestClassifier(max_depth=i, random_state=42)
    rfc.fit(x_train1, y_train1)
    y_pred2 = rfc.predict(x_test1)

    acc = accuracy_score(y_test1, y_pred2)
    acc_list.append(acc)

    f1 = f1_score(y_test1, y_pred2, average='macro')
    f1_list.append(f1)

print("=== Accuracy Score ===")
for i in acc_list:
    print(i)

print("=== F1_score ===")
for i in f1_list:
    print(i)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
acc_list2 = []
f1_list2 = []
for i in max2:
    tree = DecisionTreeClassifier(max_depth=i, random_state=42)
    tree.fit(x_train1, y_train1)
    y_pred3 = tree.predict(x_test1)
    acc = accuracy_score(y_test1, y_pred3)
    acc_list2.append(acc)

    f1 = f1_score(y_test1, y_pred3, average='macro')
    f1_list2.append(f1)


print("=== Accuracy Score ===")
for i in acc_list2:
    print(i)

print("=== F1_score ===")
for i in f1_list2:
    print(i)
```

```python
n_estimator = [1,5,10,20,50,100,200,500]
max_depth = 2
acc_list3 = []
f1_list3 = []
for i in n_estimator:
    rfc = RandomForestClassifier(max_depth=max_depth, random_state=42, n_estimators=i)
    rfc.fit(x_train1, y_train1)
    y_pred4 = rfc.predict(x_test1)

    acc = accuracy_score(y_test1, y_pred4)
    acc_list3.append(acc)

    f1 = f1_score(y_test1, y_pred4, average='macro')
    f1_list3.append(f1)

print("=== Accuracy Score ===")
for i in acc_list3:
    print(i)

print("=== F1_score ===")
for i in f1_list3:
    print(i)
```
